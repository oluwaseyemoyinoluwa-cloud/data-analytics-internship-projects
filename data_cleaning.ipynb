```python
import pandas as pd

# Create dataset
data = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 4],
    'product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Laptop'],
    'price': [800, 500, None, 800, 800],
    'quantity': [1, 2, 1, 1, 1],
    'city': ['NY', 'Chicago', 'Boston', 'NY', 'NY']
})

print("Original Data:")
print(data)

# Check missing values
print("\nMissing Values:")
print(data.isnull().sum())

# Remove missing values
data = data.dropna()

# Remove duplicates
data = data.drop_duplicates()

# Create new column
data['total_sales'] = data['price'] * data['quantity']

print("\nCleaned Data:")
print(data)

# Save cleaned data
data.to_csv('clean_data.csv', index=False)
```
