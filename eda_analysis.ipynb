```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create dataset (same as before)
data = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 4],
    'product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Laptop'],
    'price': [800, 500, 300, 800, 800],
    'quantity': [1, 2, 1, 1, 1],
    'city': ['NY', 'Chicago', 'Boston', 'NY', 'NY']
})

# Summary statistics
print("Summary Statistics:")
print(data.describe())

# Mean, Median, Mode
print("\nMean:")
print(data.mean(numeric_only=True))

print("\nMedian:")
print(data.median(numeric_only=True))

print("\nMode:")
print(data.mode())

# Histogram
plt.hist(data['price'])
plt.title("Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()

# Boxplot
sns.boxplot(x=data['price'])
plt.title("Boxplot of Price")
plt.show()

# Scatter Plot
plt.scatter(data['price'], data['quantity'])
plt.title("Price vs Quantity")
plt.xlabel("Price")
plt.ylabel("Quantity")
plt.show()

# Correlation
corr = data.corr(numeric_only=True)
print("\nCorrelation:")
print(corr)

# Heatmap
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
