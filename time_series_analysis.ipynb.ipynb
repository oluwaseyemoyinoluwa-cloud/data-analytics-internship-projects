```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Create dataset
data = pd.DataFrame({
    'price': [800, 500, 300, 800, 800]
})

# Create a date column
data['date'] = pd.date_range(start='2024-01-01', periods=len(data), freq='D')

# Set date as index
data.set_index('date', inplace=True)

# Plot time series
plt.plot(data['price'])
plt.title("Price Over Time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.show()

# Moving average
data['moving_avg'] = data['price'].rolling(window=2).mean()

plt.plot(data['price'], label='Original')
plt.plot(data['moving_avg'], label='Moving Average')
plt.legend()
plt.title("Moving Average")
plt.show()

# Decomposition
result = seasonal_decompose(data['price'], model='additive', period=2)
result.plot()
plt.show()
```
